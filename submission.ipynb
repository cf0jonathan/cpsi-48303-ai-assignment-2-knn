```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

data = pd.read_csv(url)

X = data[
    [
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
]

y = data["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=23
)

model = KNeighborsClassifier(n_neighbors=3)


```

2. The closest existing flowers to the new flower being analyzed.
3. this means that the 3 nearest neighbors are analyzed for a match.
4. The similar examples are all it has to go by when making a prediction.

```python
model.fit(X_train, y_train)
```

### Task 2
1. .fit() trains the model on the data provided.
2. the data split creates training and testing data sets, and the training portion is used here.
3. no
4. it , and verifies there aren't issues of the data just being "memorized"

```python
predictions = model.predict(X_test) 

print(predictions)

print(y_test)
```

### Task 3
1. Predict uses the just trained model to go through the test data features and predict the target.
2. The predictions and labels are identical.
3. yes
4. No

```python
accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
```

```python
model_1 = KNeighborsClassifier(n_neighbors=1)

model_1.fit(X_train, y_train)

prediction_1 = model_1.predict(X_test)

accuracy_1 = accuracy_score(y_test, prediction_1)

print(accuracy_1)


model_3 = KNeighborsClassifier(n_neighbors=3)

model_3.fit(X_train, y_train)

prediction_3 = model_3.predict(X_test)

accuracy_3 = accuracy_score(y_test, prediction_3)

print(accuracy_3)


model_5 = KNeighborsClassifier(n_neighbors=5)

model_5.fit(X_train, y_train)

prediction_5 = model_5.predict(X_test)

accuracy_5 = accuracy_score(y_test, prediction_5)

print(accuracy_5)


model_10 = KNeighborsClassifier(n_neighbors=10)

model_10.fit(X_train, y_train)

prediction_10 = model_10.predict(X_test)

accuracy_10 = accuracy_score(y_test, prediction_10)

print(accuracy_10)
```

### Complete AI Workflow

1. Real world data is collected and organized.
2. Data is inspected to understand the structure and spot patterns.
3. Graphs are created to visualize the data and look for patterns, both those we already spotted and the ones we may have missed.
4. It is determined which data is the features used for predicting, and which single aspect is the label being predicted.
5. The data is split into training and testing data.
6. We choose the ML method best fit for the task, in this case KNN.
7. The model is trained on the training data.
8. The testing data is fed into the model to test if it outputs the correct label.
9. This output is compared to the known correct label to evaluate the accuracy of the model.
10. Different parameters are used to fine tune the model in order to achieve the greatest level of accuracy.
11. We compare the accuracy of the original and refined model to see if it improved.
12. We repete steps 10 and 11 with what we've learned until we achieve the greatest possible accuracy.
